```python
us2025=data2025[(data2025["Geography"] == "Total U.S.") & (data2025["Period"] <= 32)]
us2026=data2026[(data2026["Geography"] == "Total U.S.") & (data2026["Period"] <= 32)]
```

```python
volume2025=us2025["Total Bulk and Bags"].sum()
volume2026=us2026["Total Bulk and Bags"].sum()
print(volume2025)
print(volume2026)
```

```python
volumechange=(volume2026-volume2025)/volume2025 * 100
print(volumechange)
```

```python
asp2025=((us2025["ASP Current Year"] * us2025["Total Bulk and Bags"]).sum() / us2025["Total Bulk and Bags"].sum())
asp2026=((us2026["ASP Current Year"] * us2026["Total Bulk and Bags"]).sum() / us2026["Total Bulk and Bags"].sum())
print(asp2025)
print(asp2026)
```

```python
aspchange=((asp2026-asp2025)/asp2025) * 100
print(aspchange)
```

```python
estsales2025=(volume2025*asp2025)
estsales2026=(volume2026*asp2026)
print(estsales2025)
print(estsales2026)
```

```python
estsaleschange=((estsales2026-estsales2025)/estsales2025)*100
print(estsaleschange)
```
